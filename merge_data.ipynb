```python
# 셀 1
import pandas as pd
import numpy as np
```

```python
# 셀 2: 데이터 경로 설정 및 파일 불러오기
train_a_path = './data/train/A.csv'
train_b_path = './data/train/B.csv'
train_labels_path = './data/train.csv'

train_a = pd.read_csv(train_a_path)
train_b = pd.read_csv(train_b_path)
train_labels = pd.read_csv(train_labels_path)
```

```python
# 셀 3: 데이터 병합
# train/A.csv와 train/B.csv를 상하로 연결
train_features = pd.concat([train_a, train_b])

# feature 데이터와 label 데이터를 'Test_id' 기준으로 병합
train_df = pd.merge(train_features, train_labels, on='Test_id')
```

```python
# 셀 4: 병합된 학습 데이터 확인
print("--- 최종 학습 데이터 정보 ---")
train_df.info()

print("\n\n--- 데이터 샘플 (상위 5개) ---")
display(train_df.head())
```

```python
# 셀 5: 데이터 경로 설정 및 파일 불러오기
test_a_path = './data/test/A.csv'
test_b_path = './data/test/B.csv'
test_info_path = './data/test.csv'

test_a = pd.read_csv(test_a_path)
test_b = pd.read_csv(test_b_path)
test_info = pd.read_csv(test_info_path)
```

```python
# 셀 6: 데이터 병합
# test/A.csv와 test/B.csv를 상하로 연결
test_features = pd.concat([test_a, test_b])

# feature 데이터와 test.csv의 정보를 'Test_id' 기준으로 병합
test_df = pd.merge(test_features, test_info, on='Test_id')
```

```python
# 셀 7: 병합된 테스트 데이터 확인
print("--- 최종 테스트 데이터 정보 ---")
test_df.info()

print("\n\n--- 데이터 샘플 (상위 5개) ---")
display(test_df.head())
```

```python
# 셀 8: 병합된 학습 데이터(train_df)를 CSV 파일로 저장
# 파일은 data 폴더 안에 train_merged.csv 라는 이름으로 저장됩니다.
train_df.to_csv('./data/train_merged.csv', index=False, encoding='utf-8-sig')

print("✅ './data/train_merged.csv' 파일 저장 완료!")
```

```python
# 셀 9: 병합된 테스트 데이터(test_df)를 CSV 파일로 저장
# 파일은 data 폴더 안에 test_merged.csv 라는 이름으로 저장됩니다.
test_df.to_csv('./data/test_merged.csv', index=False, encoding='utf-8-sig')

print("✅ './data/test_merged.csv' 파일 저장 완료!")
```
